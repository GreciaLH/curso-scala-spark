Ejemplos aplicados a una plataforma de música en streaming.

- case class. Una canción en una plataforma de streaming.

```scala
// DEFINICIÓN
case class Cancion(tiulo: String, artista: String, duracionSeg: Int, reproducciones: Long)

//USO
val c1 = Cancion("Bohemian Rhapsody", "Queen", 354, 2_800_000_000L)
val c2 = Cancion("Bohemian Rhapsody", "Queen", 354, 2_800_000_000L)
val c3 = Cancion("Hotel California", "Eagles", 391, 1_100_000_000L)

println(c1)          // Cancion(Bohemian Rhapsody,Queen,354,2800000000)
println(c1 == c2)    // true  ← compara por VALOR, no por referencia
println(c1 == c3)    // false

val c1Popular = c1.copy(reproducciones = 3_000_000_000L)
println(c1Popular)   // Cancion(Bohemian Rhapsody,Queen,354,3000000000)
println(c1)          // original intacto
```

## Explicación línea a línea:

- **`case class Cancion(...)`** — con una sola línea Scala genera automáticamente `toString`, `equals`, `hashCode` y `copy`. Sin `case`, habría que escribir esos cuatro métodos a mano (≈40 líneas).

- **`val c1 = Cancion(...)`** — no necesita `new`. La case class expone un método `apply` en su companion object que actúa de constructor.

- **`println(c1)`** — llama al `toString` generado, que produce `Cancion(Bohemian Rhapsody,Queen,354,2800000000)` en lugar de una dirección de memoria.

- **`c1 == c2`** — llama al `equals` generado, que compara campo a campo. Devuelve `true` porque todos los valores coinciden. Con una clase normal, habría devuelto `false` porque compararía referencias.

- **`c1.copy(reproducciones = 3_000_000_000L)`** — crea un nuevo objeto con todos los campos de `c1` salvo `reproducciones`, que toma el nuevo valor. El original `c1` no cambia: es inmutable.

- case object. Calidad de audio disponible en el streaming.

```scala
// DEFINICIÓN
case object Normal    // 128 kbps
case object Alta      // 320 kbps
case object SinConexion

// USO
val calidad = Alta
println(calidad)  // Alta  ← toString automático, no una dirección de memoria

def descripcionCalidad(c: Any): String = c match {
  case Normal      => "Calidad estándar (128 kbps)"
  case Alta        => "Alta fidelidad (320 kbps)"
  case SinConexion => "Reproducción sin conexión"
}

println(descripcionCalidad(Alta))        // Alta fidelidad (320 kbps)
println(descripcionCalidad(SinConexion)) // Reproducción sin conexión
```

## Explicación línea a línea:

- **`case object Normal`** — crea un singleton (existe una única instancia en todo el programa) con `toString` automático. Sin `case`, `println(calidad)` mostraría algo como `ammonite.$sess.cmd0$Alta$@3f7e3` en lugar de `Alta`.

- **`val calidad = Alta`** — `calidad` apunta al único objeto `Alta`. No se crea ninguna copia: todo el programa comparte la misma instancia.

- **`def descripcionCalidad(c: Any): String`** — recibe `Any` porque los tres case object no tienen ningún tipo en común declarado. Más adelante, con `sealed trait`, esto mejora.

- **`case SinConexion =>`** — el pattern matching funciona comparando si `c` es exactamente el mismo objeto singleton que `SinConexion`. Como solo existe uno, la comparación es segura.

- attern matching. Desestructuración

```scala
//Desestructuración
case class Album(titulo: String, anio: Int, numCanciones: Int)

def clasificarAlbum(a: Album): String = a match {
  case Album(_, anio, _) if anio < 1970  => "Clásico anterior al rock moderno"
  case Album(_, anio, n) if anio < 2000 && n > 15 => s"Álbum extenso de los $anio"
  case Album(t, anio, _) if anio >= 2020 => s"$t es reciente (${anio})"
  case Album(_, _, n)    if n <= 5       => "EP o single extendido"
  case _                                 => "Álbum estándar"
}

println(clasificarAlbum(Album("Abbey Road",  1969, 17)))  // Clásico anterior al rock moderno
println(clasificarAlbum(Album("OK Computer", 1997, 12)))  // Álbum estándar
println(clasificarAlbum(Album("Midnights",   2022, 13)))  // Midnights es reciente (2022)
```

## Explicación línea a línea:

- **`case Album(_, anio, _)`** — el `_` ignora campos que no necesitamos en ese caso. `anio` extrae el segundo campo directamente como variable local, sin tener que escribir `a.anio` después.

- **`if anio < 1970`** — este es el guard: el caso solo se activa si además del patrón se cumple la condición. Patrón y guard deben ser verdaderos a la vez.

- **`case Album(t, anio, _)`** — extrae dos campos a la vez con nombres propios (`t` y `anio`) para poder usarlos en la parte derecha de la flecha.

- **`case _`** — comodín de seguridad: captura cualquier `Album` que no haya encajado en ningún caso anterior.

```scala
//Matching sobre tipos
case class Cancion(titulo: String, artista: String, duracionSeg: Int, reproducciones: Long)
case class Podcast(titulo: String, programa: String, duracionSeg: Int)
case class Audiolibro(titulo: String, autor: String, capitulos: Int)

def infoContenido(c: Any): String = c match {
  case Cancion(t, a, dur, rep) =>
    f"Canción: $t — $a | ${dur / 60}min | $rep%,d escuchas"
  case Podcast(t, prog, dur) =>
    s"Podcast: $t ($prog) | ${dur / 60} min"
  case Audiolibro(t, autor, caps) =>
    s"Audiolibro: $t de $autor | $caps capítulos"
  case _ => "Contenido desconocido"
}

println(infoContenido(Cancion("Starman", "Bowie", 258, 450_000_000L)))
// Canción: Starman — Bowie | 4min | 450,000,000 escuchas
println(infoContenido(Podcast("Historia de Roma", "Dan Carlin", 14400)))
// Podcast: Historia de Roma (Dan Carlin) | 240 min
```

## Explicación línea a línea:

- **`def infoContenido(c: Any)`** — recibe `Any` porque los tres tipos no comparten jerarquía. Cada case comprueba simultáneamente el tipo real del objeto y extrae sus campos.

- **`case Cancion(t, a, dur, rep) =>`** — si `c` es una `Cancion`, extrae los cuatro campos. Si no lo es, pasa al siguiente caso. Es matching de tipo y desestructuración en un solo gesto.

- **`f"... $rep%,d ..."`** — interpola con formato: `%,d` añade separadores de miles al número de reproducciones.
