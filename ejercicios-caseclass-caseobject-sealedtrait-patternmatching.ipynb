Ejemplos aplicados a una plataforma de música en streaming.

- case class. Una canción en una plataforma de streaming.

```scala
// DEFINICIÓN
case class Cancion(tiulo: String, artista: String, duracionSeg: Int, reproducciones: Long)

//USO
val c1 = Cancion("Bohemian Rhapsody", "Queen", 354, 2_800_000_000L)
val c2 = Cancion("Bohemian Rhapsody", "Queen", 354, 2_800_000_000L)
val c3 = Cancion("Hotel California", "Eagles", 391, 1_100_000_000L)

println(c1)          // Cancion(Bohemian Rhapsody,Queen,354,2800000000)
println(c1 == c2)    // true  ← compara por VALOR, no por referencia
println(c1 == c3)    // false

val c1Popular = c1.copy(reproducciones = 3_000_000_000L)
println(c1Popular)   // Cancion(Bohemian Rhapsody,Queen,354,3000000000)
println(c1)          // original intacto
```

### Explicación línea a línea:

- **`case class Cancion(...)`** — con una sola línea Scala genera automáticamente `toString`, `equals`, `hashCode` y `copy`. Sin `case`, habría que escribir esos cuatro métodos a mano (≈40 líneas).

- **`val c1 = Cancion(...)`** — no necesita `new`. La case class expone un método `apply` en su companion object que actúa de constructor.

- **`println(c1)`** — llama al `toString` generado, que produce `Cancion(Bohemian Rhapsody,Queen,354,2800000000)` en lugar de una dirección de memoria.

- **`c1 == c2`** — llama al `equals` generado, que compara campo a campo. Devuelve `true` porque todos los valores coinciden. Con una clase normal, habría devuelto `false` porque compararía referencias.

- **`c1.copy(reproducciones = 3_000_000_000L)`** — crea un nuevo objeto con todos los campos de `c1` salvo `reproducciones`, que toma el nuevo valor. El original `c1` no cambia: es inmutable.

- case object. Calidad de audio disponible en el streaming.

```scala
// DEFINICIÓN
case object Normal    // 128 kbps
case object Alta      // 320 kbps
case object SinConexion

// USO
val calidad = Alta
println(calidad)  // Alta  ← toString automático, no una dirección de memoria

def descripcionCalidad(c: Any): String = c match {
  case Normal      => "Calidad estándar (128 kbps)"
  case Alta        => "Alta fidelidad (320 kbps)"
  case SinConexion => "Reproducción sin conexión"
}

println(descripcionCalidad(Alta))        // Alta fidelidad (320 kbps)
println(descripcionCalidad(SinConexion)) // Reproducción sin conexión
```

### Explicación línea a línea:

- **`case object Normal`** — crea un singleton (existe una única instancia en todo el programa) con `toString` automático. Sin `case`, `println(calidad)` mostraría algo como `ammonite.$sess.cmd0$Alta$@3f7e3` en lugar de `Alta`.

- **`val calidad = Alta`** — `calidad` apunta al único objeto `Alta`. No se crea ninguna copia: todo el programa comparte la misma instancia.

- **`def descripcionCalidad(c: Any): String`** — recibe `Any` porque los tres case object no tienen ningún tipo en común declarado. Más adelante, con `sealed trait`, esto mejora.

- **`case SinConexion =>`** — el pattern matching funciona comparando si `c` es exactamente el mismo objeto singleton que `SinConexion`. Como solo existe uno, la comparación es segura.

- Pattern matching

```scala
case class Cancion(titulo: String, artista: String, duracionSeg: Int, reproducciones: Long)
case class Podcast(titulo: String, programa: String, duracionSeg: Int)
case class Audiolibro(titulo: String, autor: String, capitulos: Int)

def clasificarContenido(c: Any): String = c match {
  case Cancion(t, _, _, rep) if rep > 1_000_000_000L =>
    s"Fenómeno global: '$t'"
  case Cancion(t, artista, dur, _) if dur > 420 =>
    s"Canción épica: '$t' de $artista (${dur / 60} min)"
  case Cancion(t, artista, _, _) =>
    s"Canción: '$t' de $artista"
  case Podcast(t, prog, dur) if dur > 3600 =>
    s"Maratón de podcast: '$t' ($prog) — ${dur / 3600}h ${(dur % 3600) / 60}min"
  case Podcast(t, prog, _) =>
    s"Podcast: '$t' ($prog)"
  case Audiolibro(t, autor, caps) if caps > 50 =>
    s"Audiolibro extenso: '$t' de $autor — $caps capítulos"
  case Audiolibro(t, autor, _) =>
    s"Audiolibro: '$t' de $autor"
  case _ =>
    "Contenido no reconocido"
}

val playlist: List[Any] = List(
  Cancion("Blinding Lights", "The Weeknd", 200, 3_400_000_000L),
  Cancion("Stairway to Heaven", "Led Zeppelin", 482, 400_000_000L),
  Cancion("Nube", "Artista Nuevo", 190, 800_000L),
  Podcast("Historia de Roma", "Dan Carlin", 14_400),
  Podcast("Daily Tech", "TechCast", 900),
  Audiolibro("Guerra y Paz", "Tolstoi", 78),
  Audiolibro("El Principito", "Saint-Exupéry", 8)
)

for (c <- playlist) println(clasificarContenido(c))
```

### Explicación línea a línea:

- **`def clasificarContenido(c: Any): String`** — recibe `Any` porque los tres tipos no comparten jerarquía. Cada case se encarga tanto de identificar el tipo como de extraer sus campos en un solo gesto.

- **`case Cancion(t, _, _, rep) if rep > 1_000_000_000L =>`** — aquí se combinan las tres técnicas a la vez: matching sobre el tipo `Cancion` (matching de tipo), extracción de `t` y `rep` ignorando los otros dos campos con `_` (desestructuración), y la condición sobre `rep` (guard). Solo activa este caso si el objeto es una `Cancion` con más de mil millones de reproducciones.

- **`case Cancion(t, artista, dur, _) if dur > 420 =>`** — el segundo caso para `Cancion`: si llegó aquí es que el primero no se activó, es decir, las reproducciones no superan el umbral. Ahora extrae `dur` y añade un nuevo guard sobre la duración. Los casos se evalúan en orden, por eso el más restrictivo siempre va primero.

- **`case Cancion(t, artista, _, _) =>`** — caso de escape para cualquier `Cancion` que no haya cumplido ningún guard anterior. No necesita guard porque actúa como cajón de sastre para ese tipo.

- **`case Podcast(t, prog, dur) if dur > 3600 =>`** — el `_` desaparece porque necesitamos los tres campos para construir el mensaje. El guard filtra solo los podcasts de más de una hora; el cálculo de horas y minutos se hace directamente en la interpolación.

- **`case Podcast(t, prog, _) =>`** — escape para podcasts cortos. El tercer campo no hace falta, así que se ignora con `_`.

- **`case Audiolibro(t, autor, caps) if caps > 50 =>`** — mismo patrón: desestructuración completa más guard. Un audiolibro con más de 50 capítulos merece etiqueta propia.

- **`case Audiolibro(t, autor, _) =>`** — escape para audiolibros cortos. El número de capítulos no es relevante aquí.

- **`case _ =>`** — comodín final: captura cualquier objeto que no haya encajado en ningún caso anterior. En este ejemplo nunca se activa porque cubrimos todos los tipos conocidos, pero es una buena práctica incluirlo cuando el parámetro es `Any`, ya que el compilador no puede verificar la exhaustividad sin un `sealed trait`.

- sealed trait + case class / case object. Estado de reproducción de una canción en la app.

```scala
// JERARQUÍA SELLADA
sealed trait EstadoReproduccion

case object Detenido                             extends EstadoReproduccion
case object Cargando                             extends EstadoReproduccion
case class  Reproduciendo(cancion: String,
                           posicionSeg: Int)     extends EstadoReproduccion
case class  EnPausa(cancion: String,
                    posicionSeg: Int)            extends EstadoReproduccion
case class  Error(codigo: Int, mensaje: String) extends EstadoReproduccion

// FUNCIÓN CON MATCH EXHAUSTIVO
def mostrarEstado(e: EstadoReproduccion): String = e match {
  case Detenido =>
    "⬛ Detenido"
  case Cargando =>
    "⏳ Cargando..."
  case Reproduciendo(cancion, pos) =>
    f"▶ Reproduciendo '$cancion' — ${pos / 60}%02d:${pos % 60}%02d"
  case EnPausa(cancion, pos) =>
    f"⏸ En pausa '$cancion' — ${pos / 60}%02d:${pos % 60}%02d"
  case Error(cod, msg) =>
    s"❌ Error $cod: $msg"
}

// SIMULACIÓN
val historial: List[EstadoReproduccion] = List(
  Detenido,
  Cargando,
  Reproduciendo("Bohemian Rhapsody", 0),
  Reproduciendo("Bohemian Rhapsody", 127),
  EnPausa("Bohemian Rhapsody", 200),
  Reproduciendo("Bohemian Rhapsody", 200),
  Error(503, "No se pudo conectar al servidor")
)

historial.foreach(e => println(mostrarEstado(e)))
```

### Explicación línea a línea:

- **`sealed trait EstadoReproduccion`** — declara la jerarquía. La palabra `sealed` le dice al compilador que todas las subclases están en este mismo fichero. Gracias a eso, cuando hagamos un match, el compilador sabrá exactamente cuántos casos existen y advertirá si nos olvidamos alguno.

- **`case object Detenido extends EstadoReproduccion`** — estado sin datos: la reproducción está parada. No hay información adicional que guardar, así que basta con un case object.

- **`case object Cargando extends EstadoReproduccion`** — igual que `Detenido`. Solo es un estado transitorio sin datos propios.

- **`case class Reproduciendo(cancion: String, posicionSeg: Int) extends EstadoReproduccion`** — estado con datos: necesitamos saber qué canción suena y en qué segundo va. Por eso usamos case class.

- **`case class EnPausa(cancion: String, posicionSeg: Int) extends EstadoReproduccion`** — idéntica estructura a `Reproduciendo`. La distinción semántica (pausado vs. sonando) la aporta el tipo, no un campo booleano extra. Esta es la ventaja de los ADTs frente a usar un String como estado.

- **`case class Error(codigo: Int, mensaje: String) extends EstadoReproduccion`** — los errores también tienen datos. Un case object `ErrorGenerico` no sería suficiente porque necesitamos saber el código HTTP y el mensaje descriptivo.

- **`def mostrarEstado(e: EstadoReproduccion): String`** — el compilador sabe que `EstadoReproduccion` tiene exactamente cinco subtipos. Si eliminamos el caso `Cargando`, lanza la advertencia: "match may not be exhaustive, it would fail on: Cargando". Con un String esto nunca ocurriría.

- **`case Reproduciendo(cancion, pos) =>`** — desestructura y hace matching de tipo en un solo paso: si `e` es un `Reproduciendo`, extrae sus campos como `cancion` y `pos`.

- **`f"▶ Reproduciendo '$cancion' — ${pos / 60}%02d:${pos % 60}%02d"`** — usa interpolación f-string con especificadores de formato. `%02d` formatea con cero a la izquierda para que "2:7" aparezca como "02:07".

- **`historial.foreach(e => println(mostrarEstado(e)))`** — itera la lista y aplica la función a cada estado. El compilador garantiza que ningún estado posible quedará sin tratar.
